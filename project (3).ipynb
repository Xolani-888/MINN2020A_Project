```python
def check_temperature(temperature):
    if temperature >= 100:
        return "CRITICAL"
    elif temperature >= 80:
        return "WARNING"
    else:
        return "NORMAL"


def check_vibration(vibration):
    if vibration >= 8:
        return "CRITICAL"
    elif vibration >= 6:
        return "WARNING"
    else:
        return "NORMAL"


def check_maintenance(equipment_item):
    maintenance_status = str(
        equipment_item["Maintenance_Status"]
    ).strip().upper()

    if maintenance_status == "OVERDUE":
        return "OVERDUE"
    elif maintenance_status == "DUE":
        return "DUE"
    else:
        return "NORMAL"


def monitor_all_equipment(data):
    results = []

    for index, equipment_item in data.iterrows():

        temperature = equipment_item["Temperature"]
        temperature_status = check_temperature(temperature)

        vibration = equipment_item["Vibration"]
        vibration_status = check_vibration(vibration)

        maintenance_status = check_maintenance(equipment_item)

        downtime = equipment_item["Downtime"]
        availability = equipment_item["Availability"]

        warnings = []

        if temperature_status == "WARNING":
            warnings.append("High temperature")
        elif temperature_status == "CRITICAL":
            warnings.append("Critical temperature")

        if vibration_status == "WARNING":
            warnings.append("High vibration")
        elif vibration_status == "CRITICAL":
            warnings.append("Critical vibration")

        if maintenance_status == "DUE":
            warnings.append("Maintenance due")
        elif maintenance_status == "OVERDUE":
            warnings.append("Maintenance overdue")

        if downtime > 30:
            warnings.append("Excessive downtime")

        if availability < 99.5:
            warnings.append("Low availability")

        if (temperature_status == "CRITICAL"
                or vibration_status == "CRITICAL"):
            overall_status = "CRITICAL"

        elif warnings:
            overall_status = "WARNING"

        else:
            overall_status = "NORMAL"

        results.append([
            equipment_item["Equipment_ID"],
            temperature,
            temperature_status,
            vibration,
            vibration_status,
            maintenance_status,
            downtime,
            availability,
            warnings,
            overall_status
        ])

    return results


# Monitor all equipment
results = monitor_all_equipment(equipment)

# Display the results
for result in results:
    print(result)
```

```python
def generate_equipment_alerts(data):
    alerts = []

    for index, equipment_item in data.iterrows():

        temperature_status = check_temperature(
            equipment_item["Temperature"]
        )

        vibration_status = check_vibration(
            equipment_item["Vibration"]
        )

        maintenance_status = check_maintenance(
            equipment_item
        )

        if temperature_status == "CRITICAL":
            alerts.append(
                [equipment_item["Equipment_ID"], "CRITICAL", "High temperature"]
            )

        if vibration_status == "CRITICAL":
            alerts.append(
                [equipment_item["Equipment_ID"], "CRITICAL", "High vibration"]
            )

        if maintenance_status == "OVERDUE":
            alerts.append(
                [equipment_item["Equipment_ID"], "CRITICAL", "Maintenance overdue"]
            )

        elif maintenance_status == "DUE":
            alerts.append(
                [equipment_item["Equipment_ID"], "WARNING", "Maintenance due"]
            )

    return alerts

alerts = generate_equipment_alerts(equipment)

for alert in alerts:
    print(alert)
```

```python
def classify_risk(likelihood, consequence):
    risk_score = likelihood * consequence

    if risk_score <= 4:
        risk_level = "Low"

    elif risk_score <= 9:
        risk_level = "Medium"

    elif risk_score <= 16:
        risk_level = "High"

    else:
        risk_level = "Critical"

    return risk_score, risk_level
```

```python
def calculate_likelihood(worker):
    score = 1

    if worker["PPE_Compliance"] == "No":
        score += 1

    if worker["Training_Status"] == "Incomplete":
        score += 1

    if worker["Fatigue_Level"] == "Medium":
        score += 1

    elif worker["Fatigue_Level"] == "High":
        score += 2

    if worker["Near_Misses"] >= 2:
        score += 1

    if score > 5:
        score = 5

    return score

for index, worker in workers.iterrows():
    likelihood = calculate_likelihood(worker)

    print(worker["Worker_ID"], "Likelihood:", likelihood)
```

```python
def calculate_consequence(worker):
    score = 1

    if worker["Previous_Incidents"] >= 1:
        score += 1

    if worker["Safety_Observations"] >= 4:
        score += 1

    if worker["Safety_Observations"] >= 6:
        score += 1

    if worker["Risk_Level"] == "High":
        score += 1

    elif worker["Risk_Level"] == "Critical":
        score += 2

    if score > 5:
        score = 5

    return score

for index, worker in workers.iterrows():
    consequence = calculate_consequence(worker)

    print(worker["Worker_ID"], "Consequence:", consequence)
```

```python
def assess_worker_risk(worker):
    likelihood = calculate_likelihood(worker)
    consequence = calculate_consequence(worker)

    risk_score, risk_level = classify_risk(
        likelihood,
        consequence
    )

    return likelihood, consequence, risk_score, risk_level

   
```

```python
 for index, worker in workers.iterrows():

    likelihood, consequence, risk_score, risk_level = assess_worker_risk(worker)

    print(
        worker["Worker_ID"],
        "Likelihood:", likelihood,
        "Consequence:", consequence,
        "Risk Score:", risk_score,
        "Risk Level:", risk_level
    )
```

```python
def risk_warning(risk_level):
    if risk_level == "Critical":
        return "CRITICAL RISK - Immediate attention required"

    elif risk_level == "High":
        return "HIGH RISK - Attention required"

    else:
        return "No immediate action required"

for index, worker in workers.iterrows():

    likelihood, consequence, risk_score, risk_level = assess_worker_risk(worker)

    warning = risk_warning(risk_level)

    print(
        worker["Worker_ID"],
        "| Risk Score:", risk_score,
        "| Risk Level:", risk_level,
        "|", warning
    )
```

```python
def equipment_needing_inspection(data):
    equipment_list = []

    for index, equipment_item in data.iterrows():

        temperature_status = check_temperature(
            equipment_item["Temperature"]
        )

        vibration_status = check_vibration(
            equipment_item["Vibration"]
        )

        maintenance_status = check_maintenance(
            equipment_item
        )

        if (temperature_status == "CRITICAL" or
            vibration_status == "CRITICAL" or
            maintenance_status == "OVERDUE" or
            equipment_item["Downtime"] > 30):

            equipment_list.append(
                equipment_item["Equipment_ID"]
            )

    return equipment_list

print("Equipment requiring immediate inspection:",
      equipment_needing_inspection(equipment))
```

```python
def equipment_with_most_alerts(data):
    alert_counts = {}

    for index, equipment_item in data.iterrows():

        equipment_id = equipment_item["Equipment_ID"]

        temperature_status = check_temperature(
            equipment_item["Temperature"]
        )

        vibration_status = check_vibration(
            equipment_item["Vibration"]
        )

        maintenance_status = check_maintenance(
            equipment_item
        )

        count = 0

        if temperature_status in ["WARNING", "CRITICAL"]:
            count += 1

        if vibration_status in ["WARNING", "CRITICAL"]:
            count += 1

        if maintenance_status in ["DUE", "OVERDUE"]:
            count += 1

        if equipment_item["Downtime"] > 30:
            count += 1

        if equipment_item["Availability"] < 99.5:
            count += 1

        alert_counts[equipment_id] = count

    highest = max(alert_counts.values())

    equipment = [
        equipment_id
        for equipment_id, count in alert_counts.items()
        if count == highest
    ]

    return equipment, highest

equipment_id, number = equipment_with_most_alerts(equipment)

print("Equipment with the most alerts:", equipment_id)
print("Number of alerts:", number)
```

```python
def incidents_by_shift(data):
    counts = data["Shift"].value_counts()

    day = counts.get("Day", 0)
    night = counts.get("Night", 0)

    return day, night

day, night = incidents_by_shift(incidents)

print("Day shift incidents:", day)
print("Night shift incidents:", night)
```

```python
def incident_types_by_shift(data):
    result = data.groupby(
        ["Shift", "Incident_Type"]
    ).size()

    return result

print(incident_types_by_shift(incidents))
```

```python
def management_attention(workers, incidents, equipment):

    attention = []

    high_risk = (workers["Risk_Level"].isin(
        ["High", "Critical"]
    )).sum()

    if high_risk > 0:
        attention.append(
            "High-risk workers identified"
        )

    serious_incidents = (incidents["Severity"].isin(
        ["High", "Critical"]
    )).sum()

    if serious_incidents > 0:
        attention.append(
            "High or critical safety incidents identified"
        )

    ppe = calculate_ppe_compliance(workers)

    if ppe < 90:
        attention.append(
            "PPE compliance below 90%"
        )

    inspection = equipment_needing_inspection(equipment)

    if len(inspection) > 0:
        attention.append(
            "Equipment requiring immediate inspection: "
            + str(inspection)
        )

    return attention

attention = management_attention(
    workers,
    incidents,
    equipment
)

for item in attention:
    print("-", item)
```
